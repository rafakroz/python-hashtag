# Automação de Sistemas e Processos com Python

### Desafio:

Todos os dias, o nosso sistema atualiza as vendas do dia anterior.
O seu trabalho diário, como analista, é enviar um e-mail para a diretoria, assim que começar a trabalhar, com o faturamento e a quantidade de produtos vendidos no dia anterior

E-mail da diretoria: seugmail+diretoria@gmail.com<br>
Local onde o sistema disponibiliza as vendas do dia anterior: https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga

Para resolver isso, vamos usar o pyautogui, uma biblioteca de automação de comandos do mouse e do teclado


Referência do pyautogui: https://pyautogui.readthedocs.io/en/latest/quickstart.html

```python
# Bibliotecas que serão utilizadas:
# pyautogui -> automações
# pandas -> análise de dados
# openpyxl -> manipulação de arquivos Excel

# Para instalar as bibliotecas, utilize o comando:
# pip install pyautogui pandas openpyxl
# Ou sugerido pelo ChatGPT: python -m pip install pandas numpy openpyxl pyautogui

# pyautogui.click -> Clica em um elemento na tela
# pyautogui.write -> Digita um texto na tela
# pyautogui.press -> Pressiona uma tecla no teclado
# pyautogui.hotkey -> Pressiona uma combinação de teclas no teclado
```

```python
import pyautogui as pya
import time

pya.PAUSE = 1  # Pausa de 0.5 segundos entre cada ação do pyautogui

# Passo a passo do desafio do módulo 1 do curso de Python:

# Passo 1: Entrar no sistema da empresa (link do drive)

# Abrir o Chrome
pya.press("win")
pya.write("chrome")
pya.press("enter")
time.sleep(3)  # Esperar o Chrome abrir

# Digitar o link do drive
link = "https://drive.google.com/drive/folders/149xknr9JvrlEnhNWO49zPcw0PW5icxga"
pya.write(link)

# Pressionar Enter e esperar a página carregar
pya.press("enter")
time.sleep(6)  # Esperar a página carregar
```

```python
# Passo 2: Navegar e encontrar o arquivo com a base de dados
pya.click(x=461, y=-890, clicks=2)  # Clicar duas vezes na pasta "Exportar"
time.sleep(3)  # Esperar a pasta abrir

# Passo 3: Exportar/baixar o arquivo com a base de dados para o computador
pya.click(x=461, y=-890, clicks=1, button='right')  # Clicar com o botão direito no arquivo "Vendas - Dez.xlsx"
time.sleep(3)  # Esperar o menu de contexto abrir

pya.click(x=599, y=-802, clicks=1)  # Clicar na opção "Fazer download"
time.sleep(3)  # Esperar o download iniciar
```

### Vamos agora ler o arquivo baixado para pegar os indicadores

- Faturamento
- Quantidade de Produtos

```python
import pandas as pd

# Passo 4: Calcular os indicadores (faturamento e quantidade de produtos vendidos)

# Abrir a base de dados com o pandas
caminho = r"C:\Users\rafae\Desktop\python_hashtag\modulo1\Vendas - Dez.xlsx"
tabela = pd.read_excel(caminho)

# Ver as informações da base de dados
# print(tabela)
display(tabela)

# Calcular o faturamento de todos os produtos vendidos
faturamento = tabela["Valor Final"].sum() 
print(faturamento)

# Calcular a quantidade de produtos vendidos
qtde_produtos = tabela["Quantidade"].sum()  
print(qtde_produtos)
```

### Vamos agora enviar um e-mail pelo gmail

```python
# Passo 5: Enviar os indicadores por e-mail

# Abrir uma nova aba no Chrome
pya.hotkey("ctrl", "t")

# Digitar o link do Gmail
link_gmail = "https://mail.google.com/mail"
pya.write(link_gmail)
pya.press("enter")
time.sleep(5)  # Esperar o Gmail abrir

# Clicar no botão "Escrever" para abrir a janela de novo e-mail
pya.click(x=119, y=-1080)
time.sleep(1)  # Esperar a janela de novo e-mail abrir

# Preencher os campos do e-mail destinatário
# Preencher Assunto do E-mail
# Preencher o corpo do e-mail
# Enviar o e-mail
```
