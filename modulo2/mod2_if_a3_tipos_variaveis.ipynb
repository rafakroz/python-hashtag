### Módulo 2 - Aula 3 - Tipos de Variáveis

Uma variável é um objeto, afinal, tudo em Python é um objeto.

int -> números inteiros
float -> números reais (com casas decimais, ponto flutuante)
str -> strings (cadeias de caracteres, textos)
bool ou boolean -> valores lógicos (True ou False)

Obs.: Variáveis em Python são declaradas explicitamente, mas isso não significa que você deve ficar mudando o tempo todo.

```python
faturamento = 1000
type(faturamento)
```

```python
faturamento = 1000.00
type(faturamento)
```

```python
faturamento = '1.000'
type(faturamento)
```

```python
ganha_bonus = True
type(ganha_bonus)
```
